```python
students = [
    {"name": "Ali", "marks": 85},
    {"name": "Ahmed", "marks": 60},
    {"name": "Sara", "marks": 95}
]

def add_student(name, marks):
    students.append({"name": name, "marks": marks})

add_student('Mudassar', 99)

def check_result():
    for std in students:
        if std['marks'] >= 70:
            print(std['name'], '--> Pass')
        else:
            print(std['name'], '--> Fail')

check_result()
```
