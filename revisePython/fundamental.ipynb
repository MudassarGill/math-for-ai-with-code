```python
#capstone project 

students = [
    {"name": "Ali", "age": 20, "marks": 85},
    {"name": "Ahmed", "age": 22, "marks": 60},
    {"name": "Sara", "age": 19, "marks": 95}
]

for std in students:
    print(std['name'],'-',std['marks'])
    
```

```python
for std in students:
    if std['marks']>=70:
        print(std['name'],'-',std['marks'],'Pass')
    else:
        print(std['name'], '-',std['marks'],'Fail')
```

```python
for std in students:
    if std['name']=='Ahmed':
        updated=std['marks']=75
print(students)

        
```
